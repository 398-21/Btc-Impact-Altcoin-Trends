<h2>2.1 Defining scraping and extraction functions</h2>

<p>The list of URLs to be analysed during this project is in a CSV file. Prior to importing this file and scraping the content of the URLs, a function is defined which we call to scrape the data as we iterate through the URLs. Each website we are going to scrape from has different HTML, so we also need to write a different function to extract the required information from each.

The below function checks if a webpage is accessible and returns the content if it is:</p>

```python
def getParsedWebpage(url):
    
    # Check that page is accessible for scraping
    
    headers = {"User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:66.0) Gecko/20100101 Firefox/93.0",
               "Upgrade-Insecure-Requests":"1", 
               "Accept":"text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8", 
               "Accept-Encoding":"gzip, deflate, br", 
               "Accept-Language": "en-GB,en-US;q=0.9,en;q=0.8",
               "DNT":"1"}
    
    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        soup = 'error'
    else:
        content = response.text
        
        # Parse the webpage now we've checked it's accessible
        soup = BeautifulSoup(response.text, 'html.parser')
        
    return soup
```

<h3>Extracting information for Crypto</h3>
<p>The following function extracts date and article text from a Crypto news.</p>

```python
def extractInfoCrypto(soup):
    
    # Extract article time
    date = soup.find('time').get_text()
    
    # Extract article text
    article = soup.find('div', {'class': 'caas-body'})
    
    article_text = ''
    for content in article:
        article_text = article_text + ' ' + content.get_text().lstrip().rstrip()
# removing punctuation
    article_text = article_text.replace('“', '')
    article_text = article_text.replace('”', '') 
    article_text = article_text.replace('`', '')
    article_text = article_text.replace(',', '')
    article_text = article_text.replace('.', '')
    article_text = article_text.replace('"', '')
    article_text = article_text.replace('?', '')

    return date, article_text[1:]
```

<h3>3.2.1 Identifying stop words not included in NLTK's stop words</h3>

It is possible to eliminate stop words by making use of nltk's stop words. Nevertheless, there might be additional words in the corpus that could be considered as stop words in terms of functionality and offer little value in sentiment analysis. To identify such words and eliminate them, we need to analyze the frequency of terms in each text, so we can determine which ones are the most common.[14]:

```python
# First we define a function to get the word frquencies of each corpus
# At this stage we will also remove stop words as defined by nltk's stopwords set

nltk.download('stopwords')

def getFreqs(texts):
  
    # Remove certain punctuation using regex. 
    texts = re.sub('\.|\,|\"|\“|\”|\—|\–|\-|\?|\!|\:|\;|\(|\)', '', texts)
    
    # Split into a list of words
    texts_lst = texts.lower().split()
  
    # Converting to a set means all duplicate words are removed
    unique_words_full = set(texts_lst)
    
    # Remove stop words
    unique_words = [word for word in unique_words_full if not word in stopwords.words('english')]
    
    # The dataframe within which we will store our word-frequency pairs
    word_freqs = pd.DataFrame(columns=['Word', 'Frequency'])
      
    for word in unique_words :
        word_freq_pair = {'Word': word, 'Frequency': texts_lst.count(word)}
        word_freqs = word_freqs.append(word_freq_pair, ignore_index=True)
        
    return word_freqs
```

We will use the above function to get the word frequencies:

```python
# Generate word frequencies 
BTC_word_freqs = getFreqs(Types['Full text'][0])
BNB_word_freqs = getFreqs(Types['Full text'][1])
ETH_word_freqs = getFreqs(Types['Full text'][2])
XRP_word_freqs =  getFreqs(Types['Full text'][3])
```

<h3>3.2.2 Creare filtered text</h3>
We will add in out stop words and define out stop word removal function:

```python
nltk.download('punkt')

def removeStopwords(texts):
    texts = texts.lower()
    
    # Remove certain punctuation using regex.
    texts = re.sub('\.|\,|\"|\“|\”|\—|\–|\-|\?|\!|\:|\;|\/|\(|\)', '', texts)
    
    # Tokenize
    tokens = word_tokenize(texts)
    
    # remove tokens with only punctuation, or suffixes separated from words
    punct = ["'", "''", '``', '(', ')', '%', '&', '...', '…', "‘", "’"]
    suffixes = ["'s", "n't", "'ve", "'ll", "'re", "'d"]
    remove_words = punct + suffixes + custom_stopwords + stopwords.words('english')
    
    # Create a new list with stop words removed from the corpus
    filtered_tokens = [word for word in tokens if not word in remove_words]
 
    # Reconstruct corpus as a string
    filtered_texts = ' '.join(filtered_tokens)
    
    return filtered_texts
```

```python
# Filter the editors corpora and add as a new column
Types['Filtered text']=''

for i in range(0, len(Types)):
    Types.loc[i, 'Filtered text'] = removeStopwords(Types['Full text'][i])
```

<h1>4 Sentiment analysis</h1>

For sentiment measurements, I will use nltk's sentiment analyser with the VADER lexicon. Since filtered text sometimes results in lost information due to context loss, I have decided to run the sentiment analysis using full, unfiltered text, since keeping stop words (etc.) has relatively little cost.

The sum of negative, neutral, and positive sentiments should equal one (because they represent proportions). In terms of sentiment, compound sentiment is a normalised measurement with adjusted intensity, which produces a value from 1 to -1 (positive to negative).

```python
# Download VADER (Valence Aware Dictionary and sEntiment Reasoner) lexicon, a sentiment analysis tool
nltk.download('vader_lexicon')

# Create an instance of the nltk sentiment analyser
vader = SentimentIntensityAnalyzer()
```

```python
# For each article, add the negative, neutral, positive and compound sentiment score as four new columns
Crypto['Neg_Sentiment'] = 0
Crypto['Neu_Sentiment'] = 0
Crypto['Pos_Sentiment'] = 0
Crypto['Comp_Sentiment'] = 0

for i in range(0, len(Crypto)):
    # Get sentiment score - sen_score has type dictionary
    senti_scored = vader.polarity_scores(Crypto['Text'][i])
    
    #Extract individual sentiment scores from the dictionary and enter them into the dataframe
    Crypto.loc[i,['Neg_Sentiment']] = senti_scored['neg']
    Crypto.loc[i,['Neu_Sentiment']] = senti_scored['neu']
    Crypto.loc[i,['Pos_Sentiment']] = senti_scored['pos']
    Crypto.loc[i,['Comp_Sentiment']] = senti_scored['compound']
```

Here we can see that the news is mostly neutral and there is not much variation between the positive sentiments of different types of coins. For further analysis we will visualize compound sentiment using bar plot.

<h3>5.2.2 Compound bar plot</h3>

We choose bar plot to view the date of each news article for a more accurate analysis, this allows us to know the degree of severeness of the news and thus determine the extent of its impact on the particular currency.

```python
# Use these column names
columns = ['Type', 'Date', 'Text']
# Convert the list of lists into a DataFrame
scored_news = pd.DataFrame(Crypto, columns=columns)

# Iterate through the headlines and get the polarity scores
scores = scored_news['Text'].apply(vader.polarity_scores)

# Convert the list of dicts into a DataFrame
scores_df = pd.DataFrame.from_records(scores)

# Join the DataFrames
scored_news = scored_news.join(scores_df)
scored_news
```

```python
scored_news.describe()
```

```python
# Group by date and ticker columns from scored_news and calculate the mean
mean_c = scored_news.groupby(['Date', 'Type']).mean()

# Unstack the column ticker
mean_c = mean_c.unstack('Type')

# Get the cross-section of compound in the 'columns' axis
mean_c = mean_c.xs('compound', axis='columns')
# Plot a bar chart with pandas

mean_c.plot(kind='bar', figsize=(10,5), width=1)
```

This result is expected, the majority of XRP news are in the negative range, and the other news are at about the same places as Bitcoin. In this graph, we see that apart from XRP, when BTC lies in the positive range the other 2 coins will lie within the same range for the majority of the time. We also see that ETH seems to have an extraordinarily strong correlation with BTC when compared to the other coins. This suggests that apart from XRP, BTC will have an impact on the other coins.
